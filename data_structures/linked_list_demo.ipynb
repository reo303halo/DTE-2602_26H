```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

    def print_lst(self):
        current = self
        while current:
            print(current.value)
            current = current.next

a = Node("A")
b = Node("B")
c = Node("C")

a.next = b
b.next = c

a.print_lst()
```

```python
# [A] -> [B] -> [C] -> None
# Search algorithms constantly create and explore chains of states.
```
