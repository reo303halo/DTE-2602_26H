```python
# QUEUE
# In the case of queues, we do not use a list because it is inefficient. (Could be used with .pop(0))
# Instead we do:

from collections import deque
# A plain list is O(n) for popping from the front, deque is 0(1)

queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

print(queue.popleft()) # Removes and return first item.

# Output should be "A" - FIFO - First In, First Out.
# BFS - Breadt-First Search.
```

```python
# PRIORITY QUEUE
# A normal queue will ask "Who arrived first?"
# A priority queue will ask "Who is most important?"

import heapq
# We are using heapq as it is a min-heap, so the smallest item pops first. Items are usually stored as (priority, value) tuples.
# Lowest number = highest priority

queue = []

heapq.heappush(queue, (5, "state A"))
heapq.heappush(queue, (2, "state B"))
heapq.heappush(queue, (8, "state C"))

while queue:
    priority, state = heapq.heappop(queue)
    print(priority, state)
```

```python
# Dijkstra:
# priority = distance

# A*:
# priority = distance + heuristic
```
