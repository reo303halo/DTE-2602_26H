```python
# A simple list
states = ["A", "B", "C"]
states.append("D")
print(states.pop())
```

```python
# A simple built-in Python list can represant many things, but the behavior depends on how we use it.

# Stack behavior (LIFO - Last In, First Out) which is used by DFO - Depth-First Search

stack = []

stack.append("A") # push
stack.append("B")
stack.append("C")

print(stack.pop()) # Removes and returns the last item - C
```

```python
# To prevent misuse might want to create a custom class for this:
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def empty(self):
        return len(self.items) == 0


stack = Stack()

stack.push("A")
stack.push("B")
stack.push("C")

print(stack.pop())
```

```python
stack = ["A", "B"]
stack.sort(reverse=True)
stack.insert(1, "C")
print(stack)
# This "ruins" the stack/LIFO behavior
```

```python
# The custom stack class still uses LIFO, but prevents miuse like sorting or inserting in the middle of the list like previous cell.
stack = Stack()

stack.push("A")
stack.push("B")
stack.push("C")

stack.sort(reverse=True)
```
