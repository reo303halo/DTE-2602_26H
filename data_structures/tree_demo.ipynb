```python
# A tiny tree structure demo

class TreeNode:
    def __init__(self, value):
        self.value = value
        self.children = []

root = TreeNode("Start")

root.children.append(TreeNode("New York"))
root.children.append(TreeNode("Oslo"))

print(root.children[0].value)
```

```python
left_child = root.children[0]
right_child = root.children[1]
root.children.append("Kopenhagen")

print(left_child.value)
print(right_child.value)
```

```python
class BinaryTreeNode:
    def __init__(self, value, left = None, right = None):
        self.value = value
        self.left = left
        self.right = right

root = BinaryTreeNode("Start", BinaryTreeNode("New York"), BinaryTreeNode("Oslo"))
print(root.left.value)
```
