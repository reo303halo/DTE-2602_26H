```python
from collections import deque


def bfs(graph, start, goal):
    queue = deque([start])

    visited = {start}
    parent = {start: None}
    visited_order = []

    while queue:
        node = queue.popleft() # FIFO

        # Record that we actually explored this node
        visited_order.append(node)

        if node == goal:
            break

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = node
                queue.append(neighbour)

    # Reconstruct path
    if goal not in parent:
        return visited_order, None

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    return visited_order, path
```

```python
def dfs(graph, start, goal):
    stack = [start]

    visited = {start}
    parent = {start: None}
    visited_order = []

    while stack:
        node = stack.pop() # LIFO

        visited_order.append(node)

        if node == goal:
            break

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = node
                stack.append(neighbour)

    # Reconstruct path
    if goal not in parent:
        return visited_order, None

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    return visited_order, path

        
```

```python
#        B
#       / \
#      A   D
#       \ / \
#        C   E
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "D"],
    "D": ["B", "C", "E"],
    "E": ["D", "B"]
}


bfs_path, bfs_visited = bfs(graph, "A", "E")
dfs_path, dfs_visited = dfs(graph, "A", "E")

print("BFS:")
print(bfs_path)
print(bfs_visited)

print("\nDFS:")
print(dfs_path)
print(dfs_visited)
```
